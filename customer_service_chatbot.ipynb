```python
import random
import re

print("Customer Service Chatbot")
print("Libraries imported successfully!")
```

```python
intents = {
    "greeting": {
        "patterns": ["hello", "hi", "hey", "good morning"],
        "responses": [
            "Hello! Welcome to our customer support.",
            "Hi! How can I help you today?"
        ]
    },

    "order": {
        "patterns": ["track order", "order status", "where is my order"],
        "responses": [
            "Please check your order status using your order ID on our website."
        ]
    },

    "delivery": {
        "patterns": ["delivery time", "shipping time", "when will it arrive"],
        "responses": [
            "Delivery usually takes 3–5 business days. Actual times may vary."
        ]
    },

    "return": {
        "patterns": ["return product", "return policy", "exchange item"],
        "responses": [
            "Please check our return policy or contact support to request a return."
        ]
    },

    "refund": {
        "patterns": ["refund status", "money back", "refund policy"],
        "responses": [
            "Refund processing times depend on the payment method and return status."
        ]
    },

    "payment": {
        "patterns": ["payment methods", "how can i pay", "upi", "credit card"],
        "responses": [
            "Available payment methods depend on the checkout options on our website."
        ]
    },

    "hours": {
        "patterns": ["business hours", "opening time", "support timings"],
        "responses": [
            "Please check our website for current customer support hours."
        ]
    },

    "goodbye": {
        "patterns": ["bye", "goodbye", "see you"],
        "responses": [
            "Thank you for contacting us. Have a great day!"
        ]
    }
}

print("Knowledge base created successfully!")
print("Number of intents:", len(intents))
```

```python
def preprocess(text):
    text = text.lower()
    text = re.sub(r"[^a-z0-9\s]", "", text)
    return text.split()


def get_response(user_message):
    words = preprocess(user_message)
    normalized_message = " ".join(words)

    best_intent = None
    best_score = 0

    for intent, details in intents.items():
        for pattern in details["patterns"]:
            normalized_pattern = " ".join(preprocess(pattern))

            if normalized_pattern in normalized_message:
                score = len(normalized_pattern)

                if score > best_score:
                    best_score = score
                    best_intent = intent

    if best_intent:
        return random.choice(intents[best_intent]["responses"])

    return (
        "Sorry, I didn't understand your question. "
        "Could you rephrase it or contact our support team?"
    )


print(get_response("Hello"))
print(get_response("Where is my order?"))
print(get_response("How long is delivery?"))
```

```python
print("Customer Service Chatbot")
print("Type 'exit' to stop the conversation.")

while True:
    user_input = input("You: ")

    if user_input.lower().strip() == "exit":
        print("Bot: Thank you for chatting with us!")
        break

    response = get_response(user_input)
    print("Bot:", response)
```

```python
test_cases = [
    ("Hello", "greeting"),
    ("Where is my order?", "order"),
    ("How long is delivery?", "delivery"),
    ("I want a refund", "refund"),
    ("What are your business hours?", "hours"),
]

for message, expected_intent in test_cases:
    response = get_response(message)
    print("Question:", message)
    print("Response:", response)
    print("-" * 40)
```
